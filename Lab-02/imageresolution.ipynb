## Image Resolution Catalog: Visualizing Pixel Reduction

When we decrease the resolution (pixel dimensions) of an image, we reduce the amount of visual data stored. This catalog demonstrates the visual impact of lowering the pixel count on `/content/me.jpeg`.

```python
import matplotlib.pyplot as plt
from PIL import Image

image_path = '/content/me.jpeg'

try:
    img_original = Image.open(image_path)
    width, height = img_original.size

    # Define our catalog resolutions
    resolutions = [
        {"label": "Original (100%)", "factor": 1.0, "desc": "Full detail, sharpest edges, maximum clarity."},
        {"label": "High (75%)", "factor": 0.75, "desc": "Slightly softer, but details remain mostly intact."},
        {"label": "Medium (50%)", "factor": 0.5, "desc": "Noticeable softening of fine lines and textures."},
        {"label": "Low (25%)", "factor": 0.25, "desc": "Visible pixelation starts to occur; text/fine details get blurry."},
        {"label": "Very Low (10%)", "factor": 0.1, "desc": "Heavy pixelation, blocky artifacts, loss of facial features."}
    ]

    # Generate and plot each catalog item
    for item in resolutions:
        factor = item["factor"]
        new_w = max(1, int(width * factor))
        new_h = max(1, int(height * factor))

        # Resize image
        resized = img_original.resize((new_w, new_h), Image.Resampling.LANCZOS)

        # Create a display figure to clearly showcase the resolution details
        fig, ax = plt.subplots(figsize=(6, 8))
        ax.imshow(resized)
        ax.set_title(f"{item['label']} - {new_w}x{new_h} pixels", fontsize=14, fontweight='bold')
        ax.axis('off')
        plt.show()

        print(f"Description: {item['desc']}")
        print("-" * 80)

except FileNotFoundError:
    print(f"Error: The file '{image_path}' was not found.")
except Exception as e:
    print(f"An error occurred: {e}")
```
